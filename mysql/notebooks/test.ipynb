# Creating database

```python
%%sql
# list all the available databases (folders where to put the tables)
SHOW DATABASES;
```

```python
%%sql
# create a new database before creating a table inside
CREATE DATABASE my_db;
# select that database so that every created table is created inside that database
USE my_db;
# show again the list of the databases
SHOW DATABASES;
```

```python
%%sql
# if we want to remove a database
DROP DATABASE my_db;
```

# Creating tables

```python
%%sql
# create a new table in the selected database named "exams" with 2 columns
CREATE TABLE exams (
    name varchar(60) NOT NULL, # this value cannot be NULL
    professor varchar(20) DEFAULT 'N.A.',
    difficulty int UNIQUE, # UNIQUE: different rows with the same value under 'difficulty' are not allowed
    PRIMARY KEY(name) # specify a column that should be both NOT NULL and UNIQUE
);
# list all the tables created
SHOW TABLES;
```

```python
%%sql
# list all the features (columns) of a specific table
DESCRIBE exams;
```

# Inserting data

```python
%%sql
# modify a table by removing a column
ALTER TABLE exams DROP difficulty;
# modify a table by adding a new column with a specific type
ALTER TABLE exams ADD days_left INT;

DESCRIBE exams;
```

```python
%%sql
CREATE TABLE pushups(
    day INT AUTO_INCREMENT, # this value gets incremented every time new values are added
    series INT,
    total INT DEFAULT 100,
    PRIMARY KEY(day)
);
```
